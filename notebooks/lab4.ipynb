```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
print(a)
print("Array ka size:", a.size)
print("Array ka shape:", a.shape)
```

```python
print("Dimensions:", a.ndim)
print("Data type:", a.dtype)

print("Zeros:", np.zeros(5))
print("Ones:", np.ones(5))

print("Range:", np.arange(1, 10))
print("Equal spacing:", np.linspace(0, 1, 5))
```

```python
a = np.array([10, 20, 30, 40, 50])

print("Pehla element:", a[0])
print("Teesra element:", a[2])
print("Aakhri element:", a[-1])
print("Pehle teen elements:", a[0:3])
print("Doosre se chauthe tak:", a[1:4])
```

```python
a = np.arange(12)

print(a.reshape(3, 4))
print("Nayi shape:", a.reshape(2, -1).shape)

print(np.array([1, 2, 3]).reshape(3, 1))
```

```python
a = np.arange(6)
print(a.reshape(6, 1))

scores = np.array([[80, 70, 90, 60],
                   [65, 75, 85, 95],
                   [90, 88, 76, 82]])

print("Scores ki shape:", scores.shape)
print("Pehle student ke marks:", scores[0])
```

```python
bonus = np.array([5, 0, 10, 0])
print("Bonus ke baad marks:")
print(scores + bonus)

print("Har student ka average:")
print(scores.mean(axis=1))

print("Har subject ka average:")
print(scores.mean(axis=0))
```

```python
print("20 mein se marks:")
print(scores / 5)

print("Har student ka total:")
print(scores.sum(axis=1))

print("60 ya zyada marks:")
print((scores >= 60).sum())
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[10, 20], [30, 40]])

print("Element-wise multiplication:")
print(A * B)

print("Matrix multiplication:")
print(A @ B)

print("Transpose:")
print(A.T)
```

```python
rng = np.random.default_rng(7)
marks = rng.integers(0, 101, size=(5, 3))

print(marks)
print("Har subject ka average:", marks.mean(axis=0))
```
