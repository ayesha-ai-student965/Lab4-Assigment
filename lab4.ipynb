```python
import numpy as np

scores = np.array([[45, 78, 90, 60],
                   [32, 65, 70, 55],
                   [88, 92, 79, 95]])

print(scores.ndim, scores.shape, scores.size, scores.dtype)
print(np.zeros((2, 3)))
print(np.arange(0, 10, 2), np.linspace(0, 1, 5))

lst = [1, 2, 3]
arr = np.array(lst)
print(lst * 2, arr * 2)
```

```python
print(np.ones((2, 3)))
print(np.full((2, 2), 7))

a = np.zeros((2, 3, 4))
print(a.ndim)
print(a.shape)
print(a.size)

print(np.array([1, 2.5, 3]).dtype)
```

```python
bonus = np.array([5, 0, 10, 0])
print(scores + bonus)

col = np.array([[1], [2], [3]])
print(col + np.array([10, 20, 30]))

means = scores.mean(axis=0)
print(np.round(scores - means, 1))
```

```python
marks_list = [40, 60, 72]

with_loop = []
for m in marks_list:
    with_loop.append(m / 80 * 100)

vectorized = np.array(marks_list) / 80 * 100
print(with_loop, vectorized)

print(scores.sum(), scores.mean(), scores.min(), scores.max())
print(scores.sum(axis=0))
print(scores.sum(axis=1))
print(np.round(scores.std(), 2), np.round(scores.std(ddof=1), 2))
print(scores >= 60)
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[10, 20], [30, 40]])

print(A * B)
print(A @ B)
print(A.T)

X = np.array([[1, 2, 3], [4, 5, 6]])
print((X @ X.T).shape)
```

```python
M = np.array([[2, 1], [1, 3]])
rhs = np.array([5, 10])

sol = np.linalg.solve(M, rhs)

print(sol)
print(np.linalg.det(M))
print(np.allclose(M @ sol, rhs))
print(np.linalg.inv(M))
print(np.linalg.norm(np.array([3, 4])))
```
